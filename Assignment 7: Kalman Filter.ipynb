#**Analytical Derivation**

**1. Derivation of $x_k^- \sim \mathcal{N}(m_k^-, P_k^-)$**<br>
Given the system process equation:
$$x_k^- = A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}$$

Where $x_{k-1}^+ \sim \mathcal{N}(m_{k-1}, P_{k-1})$ and $w_{k-1} \sim \mathcal{N}(0, \Sigma_p)$ are independent Gaussian random variables.

**Step 1.1: Calculate Expected Value (Mean)**

$$m_k^- = \mathbb{E}[x_k^-]$$
$$m_k^- = \mathbb{E}[A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}]$$
$$m_k^- = A_{k-1} \mathbb{E}[x_{k-1}^+] + G_{k-1} \mathbb{E}[w_{k-1}]$$

Substitute $\mathbb{E}[x_{k-1}^+] = m_{k-1}$ and $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = A_{k-1} m_{k-1} + G_{k-1}(0)$$
$$m_k^- = A_{k-1} m_{k-1}$$

**Step 1.2: Calculate Covariance**

Using the covariance definition $P_k^- = \mathbb{E}[(x_k^- - m_k^-)(x_k^- - m_k^-)^T]$:
$$x_k^- - m_k^- = (A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}) - A_{k-1} m_{k-1}$$
$$x_k^- - m_k^- = A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1} w_{k-1}$$

$$P_k^- = \mathbb{E}\left[ \left( A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1} w_{k-1} \right) \left( A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1} w_{k-1} \right)^T \right]$$

$$P_k^- = \mathbb{E}\left[ A_{k-1}(x_{k-1}^+ - m_{k-1})(x_{k-1}^+ - m_{k-1})^T A_{k-1}^T \right] + \mathbb{E}\left[ G_{k-1} w_{k-1} w_{k-1}^T G_{k-1}^T \right] + \dots$$
$$\dots + \mathbb{E}\left[ A_{k-1}(x_{k-1}^+ - m_{k-1}) w_{k-1}^T G_{k-1}^T \right] + \mathbb{E}\left[ G_{k-1} w_{k-1} (x_{k-1}^+ - m_{k-1})^T A_{k-1}^T \right]$$

Since $x_{k-1}^+$ and $w_{k-1}$ are independent, their cross-covariance expectations are zero:$$\mathbb{E}[(x_{k-1}^+ - m_{k-1}) w_{k-1}^T] = 0$$

Bringing the deterministic matrices outside the expectations:
$$P_k^- = A_{k-1} \mathbb{E}[(x_{k-1}^+ - m_{k-1})(x_{k-1}^+ - m_{k-1})^T] A_{k-1}^T + G_{k-1} \mathbb{E}[w_{k-1} w_{k-1}^T] G_{k-1}^T$$
Substitute $\text{Var}(x_{k-1}^+) = P_{k-1}$ and $\text{Var}(w_{k-1}) = \Sigma_p$:
$$P_k^- = A_{k-1} P_{k-1} A_{k-1}^T + G_{k-1} \Sigma_p G_{k-1}^T$$

**2. Derivation of $y_k^- \sim \mathcal{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$**

Given the measurement model:
$$y_k^- = H_k x_k^- + z_k$$
Where $x_k^- \sim \mathcal{N}(m_k^-, P_k^-)$ and $z_k \sim \mathcal{N}(0, \Sigma_m)$ are independent.

**Step 2.1: Mean of $y_k^-$**
$$\mathbb{E}[y_k^-] = \mathbb{E}[H_k x_k^- + z_k]$$
$$\mathbb{E}[y_k^-] = H_k \mathbb{E}[x_k^-] + \mathbb{E}[z_k]$$
$$\mathbb{E}[y_k^-] = H_k m_k^- + 0 = H_k m_k^-$$

**Step 2.2: Covariance of $y_k^-$**
$$\text{Var}(y_k^-) = \mathbb{E}[(y_k^- - \mathbb{E}[y_k^-])(y_k^- - \mathbb{E}[y_k^-])^T]$$
$$y_k^- - H_k m_k^- = H_k(x_k^- - m_k^-) + z_k$$

$$\text{Var}(y_k^-) = \mathbb{E}\left[ \left( H_k(x_k^- - m_k^-) + z_k \right) \left( H_k(x_k^- - m_k^-) + z_k \right)^T \right]$$

$$\text{Var}(y_k^-) = H_k \mathbb{E}[(x_k^- - m_k^-)(x_k^- - m_k^-)^T] H_k^T + \mathbb{E}[z_k z_k^T] + H_k \mathbb{E}[(x_k^- - m_k^-)z_k^T] + \mathbb{E}[z_k(x_k^- - m_k^-)^T]H_k^T$$

Since $x_k^-$ and $z_k$ are uncorrelated, cross terms evaluate to zero:
$$\text{Var}(y_k^-) = H_k P_k^- H_k^T + \Sigma_m$$



**3. Derivation of the Joint Distribution**

To create the joint Gaussian state-measurement matrix, we need to solve for the cross-covariance term $\text{Cov}(x_k^-, y_k^-)$.

**Step 3.1: Compute Cross-Covariance**
$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-)(y_k^- - H_k m_k^-)^T \right]$$\

Substitute the definition $(y_k^- - H_k m_k^-) = H_k(x_k^- - m_k^-) + z_k$:$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-) \left( H_k(x_k^- - m_k^-) + z_k \right)^T \right]$$$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-)(x_k^- - m_k^-)^T H_k^T + (x_k^- - m_k^-)z_k^T \right]$$

$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-)(x_k^- - m_k^-)^T \right] H_k^T + \mathbb{E}\left[ (x_k^- - m_k^-)z_k^T \right]$$

Substitute $\mathbb{E}\left[ (x_k^- - m_k^-)(x_k^- - m_k^-)^T \right] = P_k^-$ and $\mathbb{E}\left[ (x_k^- - m_k^-)z_k^T \right] = 0$:
$$\text{Cov}(x_k^-, y_k^-) = P_k^- H_k^T$$
By property of covariance transposition, $\text{Cov}(y_k^-, x_k^-) = \left(\text{Cov}(x_k^-, y_k^-)\right)^T$:
$$\text{Cov}(y_k^-, x_k^-) = (P_k^- H_k^T)^T = H_k P_k^-$$

**Step 3.2: Assemble Joint Vector**
Putting together the individual means and blocks of the computed covariance matrices yields:
$$\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix}, \begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix} \right)$$

**4 & 5. Conditional Update (Kalman Filter Correction Step)**

Given a joint Gaussian distribution:
$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} \mu_X \\ \mu_Y \end{bmatrix}, \begin{bmatrix} \Sigma_{XX} & \Sigma_{XY} \\ \Sigma_{YX} & \Sigma_{YY} \end{bmatrix} \right)$$
The standard Lemma for Gaussian conditioning dictates:
$$\mathbb{E}[X \mid Y = y] = \mu_X + \Sigma_{XY} \Sigma_{YY}^{-1}(y - \mu_Y)$$$$\text{Var}(X \mid Y = y) = \Sigma_{XX} - \Sigma_{XY} \Sigma_{YY}^{-1} \Sigma_{YX}$$

**Step 4/5.1: Map the Joint Variables**
From the matrix in part 3:

$\mu_X = m_k^-$ <br>
$\mu_Y = H_k m_k^-$<br>
$\Sigma_{XX} = P_k^-$<br>
$\Sigma_{XY} = P_k^- H_k^T$$\Sigma_{YX} = H_k P_k^-$<br>
$\Sigma_{YY} = H_k P_k^- H_k^T + \Sigma_m$<br>

**Step 4/5.2: Define Kalman Gain ($K_k$)**

Let $K_k = \Sigma_{XY} \Sigma_{YY}^{-1}$:
$$K_k = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$
**Step 4/5.3: Calculate Conditional Mean** $$\mathbb{E}[x_k^- \mid y_k^- = y_k^{\text{obs}}] = m_k^- + \Sigma_{XY} \Sigma_{YY}^{-1} (y_k^{\text{obs}} - \mu_Y)$$
$$\mathbb{E}[x_k^- \mid y_k^- = y_k^{\text{obs}}] = m_k^- + K_k (y_k^{\text{obs}} - H_k m_k^-)$$
**Step 4/5.4: Calculate Conditional Covariance**
$$\text{Var}(x_k^- \mid y_k^- = y_k^{\text{obs}}) = \Sigma_{XX} - \Sigma_{XY} \Sigma_{YY}^{-1} \Sigma_{YX}$$
$$\text{Var}(x_k^- \mid y_k^- = y_k^{\text{obs}}) = P_k^- - K_k (H_k P_k^-)$$
Factor out $P_k^-$ from the right-hand side:
$$\text{Var}(x_k^- \mid y_k^- = y_k^{\text{obs}}) = (I - K_k H_k) P_k^-$$

**Step 4/5.5: Final Match to Question 4 Definitions**
By setting $x_k^+ \triangleq (x_k^- \mid y_k^- = y_k^{\text{obs}})$, it follows a Gaussian profile $\mathcal{N}(m_k, P_k)$ matching the expressions derived directly above:
$$m_k = m_k^- + K_k(y_k^{\text{obs}} - H_k m_k^-)$$
$$P_k = (I - K_k H_k)P_k^-$$

#**1-D Example**


**1. Derivation of $m_k^-$ and $P_k^-$ (Prediction Step)**

Given the 1-D scalar process equation:
$$x_k^- = a x_{k-1}^+ + w_{k-1}$$
Where $x_{k-1}^+ \sim \mathcal{N}(m_{k-1}, P_{k-1})$ and $w_{k-1} \sim \mathcal{N}(0, q)$ are independent scalar random variables.<br>
Expected Value (Mean $m_k^-$):
$$m_k^- = \mathbb{E}[x_k^-] = \mathbb{E}[a x_{k-1}^+ + w_{k-1}]$$
Using the linearity of expectation:
$$m_k^- = a \mathbb{E}[x_{k-1}^+] + \mathbb{E}[w_{k-1}]$$
Substitute $\mathbb{E}[x_{k-1}^+] = m_{k-1}$ and $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = a m_{k-1}$$Variance ($P_k^-$):$$P_k^- = \text{Var}(x_k^-) = \text{Var}(a x_{k-1}^+ + w_{k-1})$$
Since $x_{k-1}^+$ and $w_{k-1}$ are independent, their variances sum linearly:
$$P_k^- = a^2 \text{Var}(x_{k-1}^+) + \text{Var}(w_{k-1})$$
Substitute $\text{Var}(x_{k-1}^+) = P_{k-1}$ and $\text{Var}(w_{k-1}) = q$:
$$P_k^- = a^2 P_{k-1} + q$$


**2. Derivation of $m_k$ and $P_k$ (Update Step)**

From the scalar joint Gaussian setup between state $x_k^-$ and measurement $y_k^- = h x_k^- + z_k$ (where $z_k \sim \mathcal{N}(0,r)$), the conditional distribution $p(x_k^- \mid y_k^- = y_k^{\text{obs}})$ uses the scalar Gaussian conditioning identities.<br>

First, define the innovation covariance $S_k$:

$$S_k = \text{Var}(y_k^-) = \text{Var}(h x_k^- + z_k) = h^2 P_k^- + r$$

Next, define the scalar Kalman Gain $K_k$:
$$K_k = \frac{\text{Cov}(x_k^-, y_k^-)}{\text{Var}(y_k^-)} = \frac{P_k^- h}{S_k}$$

Mean ($m_k$):
$$m_k = m_k^- + K_k v_k$$

Where $v_k = y_k^{\text{obs}} - h m_k^-$ is the innovation residual. Substituting $K_k$:
$$m_k = m_k^- + \frac{P_k^- h}{S_k} (y_k^{\text{obs}} - h m_k^-)$$

Variance ($P_k$):
$$P_k = (1 - K_k h) P_k^-$$

Substituting $K_k$:
$$P_k = \left(1 - \frac{P_k^- h}{S_k} \cdot h\right) P_k^- = \left(1 - \frac{P_k^- h^2}{S_k}\right) P_k^-$$

**3. Predictive Measurement Distribution $p(y_k^- \mid Y_{k-1})$**

We want the distribution of $y_k^- = h x_k^- + z_k$ given all observations up to time $k-1$. This depends directly on the predicted state $x_k^- \mid Y_{k-1} \sim \mathcal{N}(m_k^-, P_k^-)$.

Mean:
$$\mathbb{E}[y_k^- \mid Y_{k-1}] = \mathbb{E}[h x_k^- + z_k \mid Y_{k-1}]$$$$\mathbb{E}[y_k^- \mid Y_{k-1}] = h \mathbb{E}[x_k^- \mid Y_{k-1}] + \mathbb{E}[z_k] = h m_k^-$$
Variance:
$$\text{Var}(y_k^- \mid Y_{k-1}) = \text{Var}(h x_k^- + z_k \mid Y_{k-1})$$

Since $z_k$ is independent of the past states:
$$\text{Var}(y_k^- \mid Y_{k-1}) = h^2 \text{Var}(x_k^- \mid Y_{k-1}) + \text{Var}(z_k) = h^2 P_k^- + r$$
Thus:
$$p(y_k^- \mid Y_{k-1}) = \mathcal{N}(h m_k^-, h^2 P_k^- + r)$$

**4. Posterior-Predictive Measurement Distribution $p(y_k^- \mid Y_k)$**<br>
This distribution represents the model's measurement space after updating the state distribution with the current observation $y_k^{\text{obs}}$, mapping $x_k^+ \mid Y_k \sim \mathcal{N}(m_k, P_k)$ back through the measurement equation $y_k^- = h x_k^+ + z_k$.

Mean:
$$\mathbb{E}[y_k^- \mid Y_k] = \mathbb{E}[h x_k^+ + z_k \mid Y_k] = h m_k + 0 = h m_k$$
Variance:
$$\text{Var}(y_k^- \mid Y_k) = \text{Var}(h x_k^+ + z_k \mid Y_k) = h^2 P_k + r$$
Thus:
$$p(y_k^- \mid Y_k) = \mathcal{N}(h m_k, h^2 P_k + r)$$

**5. Conceptual Evolution of Distributions**

To visualize how the distributions evolve from the Prior (before update) to the Posterior (after update), consider typical numerical values where the measurement narrows down state uncertainty:

*   **Parameters:** Let $a = 1$, $h = 1$, $q = 1$, $r = 0.5$.
*   **Prior Stage:** The state prediction profile $p(x_k \mid Y_{k-1})$ has a broader variance $P_k^-$.
*   **Posterior Stage:** Incorporating a highly certain measurement observation collapses the variance profile so that $P_k < P_k^-$, shifting the mean $m_k$ closer to the observed measurement value.


#**2D-Position Estimation**

##**PART A**

**1. Derivation of the State Transition Matrix ($A$)**<br>
Under a constant-velocity kinematic model with a time sampling interval of $\Delta t$, the continuous equations of motion for both orthogonal 2D dimensions ($x$ and $y$) are given by Newton's basic kinematic laws:$$\begin{aligned}
p_x(k) &= p_x(k-1) + v_x(k-1)\Delta t \\
p_y(k) &= p_y(k-1) + v_y(k-1)\Delta t \\
v_x(k) &= v_x(k-1) \\
v_y(k) &= v_y(k-1)
\end{aligned}$$

Writing this unforced linear system in vector-matrix form $x_k = A x_{k-1}$:

$$\begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix} \begin{bmatrix} p_x(k-1) \\ p_y(k-1) \\ v_x(k-1) \\ v_y(k-1) \end{bmatrix}$$

Extracting the coefficient matrix yields:
$$A = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix}$$

**2. Derivation of the Measurement Matrix ($H$)**

The problem specifies that the measurement vector $y_k$ contains only the two position coordinates:
$$y_k = \begin{bmatrix} p_x^{\text{meas}}(k) \\ p_y^{\text{meas}}(k) \end{bmatrix}$$

We set up the linear measurement observation relationship $y_k = H x_k$:

$$\begin{bmatrix} p_x^{\text{meas}}(k) \\ p_y^{\text{meas}}(k) \end{bmatrix} = H \begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix}$$

To extract exactly $p_x(k)$ for the first row and $p_y(k)$ for the second row while totally ignoring the velocity components:

$$H = \begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0 \end{bmatrix}$$

**3. Derivation of the Noise Coupling Matrix ($G$)**

When random acceleration acts as the process noise input, let $w_{k-1} = \begin{bmatrix} a_x \\ a_y \end{bmatrix}$ be the 2D acceleration noise vector.

Applying the classical kinematics expansion where a constant acceleration over interval $\Delta t$ updates position by $\frac{1}{2}a\Delta t^2$ and velocity by $a\Delta t$:

$$\begin{aligned}
p_x(k) &= \dots + \frac{1}{2}\Delta t^2 a_x \\
p_y(k) &= \dots + \frac{1}{2}\Delta t^2 a_y \\
v_x(k) &= \dots + \Delta t a_x \\
v_y(k) &= \dots + \Delta t a_y
\end{aligned}$$

Writing only the noise contribution block in the state space formula form $G w_{k-1}$:

$$\begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix}_{\text{noise}} = \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix} \begin{bmatrix} a_x \\ a_y \end{bmatrix}$$

Thus, the process noise gain matrix is explicitly verified as:
$$G = \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix}$$

##**PART B**

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio
from scipy.linalg import block_diag

class KalmanFilter2D:
    """
    2D Kalman Filter for position estimation with constant velocity model
    """
    def __init__(self, dt=1.0, process_noise_std=0.5, measurement_noise_std=10.0):
        """
        Initialize Kalman Filter

        Parameters:
        dt: time step
        process_noise_std: standard deviation of process noise (acceleration)
        measurement_noise_std: standard deviation of measurement noise
        """
        self.dt = dt

        # State transition matrix A
        self.A = np.array([
            [1, 0, dt, 0],
            [0, 1, 0, dt],
            [0, 0, 1, 0],
            [0, 0, 0, 1]
        ])

        # Measurement matrix H
        self.H = np.array([
            [1, 0, 0, 0],
            [0, 1, 0, 0]
        ])

        # Process noise matrix G
        self.G = np.array([
            [0.5 * dt**2, 0],
            [0, 0.5 * dt**2],
            [dt, 0],
            [0, dt]
        ])

        # Process noise covariance (acceleration noise)
        self.process_noise_cov = np.array([
            [process_noise_std**2, 0],
            [0, process_noise_std**2]
        ])

        # Measurement noise covariance
        self.measurement_noise_cov = np.array([
            [measurement_noise_std**2, 0],
            [0, measurement_noise_std**2]
        ])

        # Process noise covariance in state space
        self.Q = self.G @ self.process_noise_cov @ self.G.T

        # Initial state and covariance
        self.x = None  # State vector [px, py, vx, vy]
        self.P = None  # State covariance matrix

    def initialize(self, initial_position, initial_velocity=None, position_uncertainty=100.0, velocity_uncertainty=10.0):
        """
        Initialize the filter with initial estimates

        Parameters:
        initial_position: [px, py]
        initial_velocity: [vx, vy] (optional)
        position_uncertainty: initial position uncertainty
        velocity_uncertainty: initial velocity uncertainty
        """
        if initial_velocity is None:
            initial_velocity = [0, 0]

        self.x = np.array([initial_position[0], initial_position[1],
                          initial_velocity[0], initial_velocity[1]])

        self.P = np.array([
            [position_uncertainty, 0, 0, 0],
            [0, position_uncertainty, 0, 0],
            [0, 0, velocity_uncertainty, 0],
            [0, 0, 0, velocity_uncertainty]
        ])

    def predict(self):
        """Prediction step"""
        if self.x is None:
            raise ValueError("Filter not initialized. Call initialize() first.")

        # Predict state
        self.x = self.A @ self.x

        # Predict covariance
        self.P = self.A @ self.P @ self.A.T + self.Q

        return self.x[:2].copy()

    def update(self, measurement):
        """
        Update step with measurement

        Parameters:
        measurement: [px_meas, py_meas]

        Returns:
        updated position estimate
        """
        if self.x is None:
            raise ValueError("Filter not initialized. Call initialize() first.")

        # Convert measurement to numpy array
        z = np.array(measurement)

        # Innovation (residual)
        y = z - self.H @ self.x

        # Innovation covariance
        S = self.H @ self.P @ self.H.T + self.measurement_noise_cov

        # Kalman gain
        K = self.P @ self.H.T @ np.linalg.inv(S)

        # Update state
        self.x = self.x + K @ y

        # Update covariance (Joseph form for numerical stability)
        I = np.eye(4)
        self.P = (I - K @ self.H) @ self.P @ (I - K @ self.H).T + K @ self.measurement_noise_cov @ K.T

        return self.x[:2].copy()

    def get_state(self):
        """Get current state estimate"""
        return self.x.copy()

    def get_covariance(self):
        """Get current covariance matrix"""
        return self.P.copy()

def generate_gps_data(true_trajectory, measurement_noise_std=10.0, dt=1.0):
    """
    Generate synthetic GPS measurements

    Parameters:
    true_trajectory: array of true positions and velocities [px, py, vx, vy] at each timestep
    measurement_noise_std: standard deviation of measurement noise

    Returns:
    measurements with added noise
    """
    measurements = []
    for state in true_trajectory:
        # Add Gaussian noise to position measurements
        noise = np.random.normal(0, measurement_noise_std, 2)
        measurement = state[:2] + noise
        measurements.append(measurement)
    return np.array(measurements)

def run_kalman_filter(measurements, dt=1.0, process_noise_std=0.5, measurement_noise_std=10.0):
    """
    Run Kalman filter on measurements

    Returns:
    filtered_states: array of filtered states [px, py, vx, vy]
    filtered_positions: array of filtered positions [px, py]
    """
    # Initialize filter
    kf = KalmanFilter2D(dt=dt,
                        process_noise_std=process_noise_std,
                        measurement_noise_std=measurement_noise_std)

    # Initialize with first measurement and zero velocity
    kf.initialize(measurements[0], initial_velocity=[0, 0])

    filtered_states = []
    filtered_positions = []

    # Process all measurements
    for measurement in measurements:
        kf.predict()
        position = kf.update(measurement)
        filtered_states.append(kf.get_state())
        filtered_positions.append(position)

    return np.array(filtered_states), np.array(filtered_positions)

def create_comparison_plot(measurements, true_trajectory, filtered_positions,
                          covariance_history=None, title="Kalman Filter Results"):
    """
    Create comprehensive visualization using Plotly

    Parameters:
    measurements: noisy GPS measurements
    true_trajectory: true positions and velocities
    filtered_positions: Kalman filtered positions
    covariance_history: list of covariance matrices for uncertainty visualization
    """
    # Create subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=("Position Estimate", "Velocity Estimate",
                       "Position Error", "Measurement vs Filtered"),
        specs=[[{"secondary_y": False}, {"secondary_y": False}],
               [{"secondary_y": False}, {"secondary_y": False}]]
    )

    time_steps = np.arange(len(true_trajectory))

    # 1. Position Estimate
    fig.add_trace(
        go.Scatter(x=time_steps, y=true_trajectory[:, 0],
                  mode='lines', name='True X', line=dict(color='green', width=3)),
        row=1, col=1
    )
    fig.add_trace(
        go.Scatter(x=time_steps, y=true_trajectory[:, 1],
                  mode='lines', name='True Y', line=dict(color='blue', width=3)),
        row=1, col=1
    )
    fig.add_trace(
        go.Scatter(x=time_steps, y=filtered_positions[:, 0],
                  mode='lines+markers', name='Filtered X',
                  line=dict(color='red', dash='dash'), marker=dict(size=4)),
        row=1, col=1
    )
    fig.add_trace(
        go.Scatter(x=time_steps, y=filtered_positions[:, 1],
                  mode='lines+markers', name='Filtered Y',
                  line=dict(color='orange', dash='dash'), marker=dict(size=4)),
        row=1, col=1
    )

    # 2. Velocity Estimate
    filtered_velocities = filtered_positions[:, 0]  # Placeholder - actually need velocity states
    fig.add_trace(
        go.Scatter(x=time_steps, y=true_trajectory[:, 2],
                  mode='lines', name='True Vx', line=dict(color='green', width=3)),
        row=1, col=2
    )
    fig.add_trace(
        go.Scatter(x=time_steps, y=true_trajectory[:, 3],
                  mode='lines', name='True Vy', line=dict(color='blue', width=3)),
        row=1, col=2
    )

    # 3. Position Error (RMSE)
    rmse_x = np.sqrt(np.mean((filtered_positions[:, 0] - true_trajectory[:, 0])**2))
    rmse_y = np.sqrt(np.mean((filtered_positions[:, 1] - true_trajectory[:, 1])**2))

    fig.add_trace(
        go.Scatter(x=time_steps, y=true_trajectory[:, 0] - filtered_positions[:, 0],
                  mode='lines', name='Error X', line=dict(color='red')),
        row=2, col=1
    )
    fig.add_trace(
        go.Scatter(x=time_steps, y=true_trajectory[:, 1] - filtered_positions[:, 1],
                  mode='lines', name='Error Y', line=dict(color='orange')),
        row=2, col=1
    )
    fig.add_hline(y=0, line_dash="dash", line_color="gray", row=2, col=1)

    # 4. Measurement vs Filtered (2D trajectory plot)
    fig.add_trace(
        go.Scatter(x=measurements[:, 0], y=measurements[:, 1],
                  mode='markers', name='GPS Measurements',
                  marker=dict(color='gray', size=6, opacity=0.5)),
        row=2, col=2
    )
    fig.add_trace(
        go.Scatter(x=true_trajectory[:, 0], y=true_trajectory[:, 1],
                  mode='lines+markers', name='True Trajectory',
                  line=dict(color='green', width=3), marker=dict(size=8)),
        row=2, col=2
    )
    fig.add_trace(
        go.Scatter(x=filtered_positions[:, 0], y=filtered_positions[:, 1],
                  mode='lines+markers', name='Filtered Trajectory',
                  line=dict(color='red', width=2), marker=dict(size=6)),
        row=2, col=2
    )

    # Update layout
    fig.update_layout(
        height=800,
        showlegend=True,
        title_text=f"{title}<br>RMSE X: {rmse_x:.2f}, RMSE Y: {rmse_y:.2f}",
        title_font_size=16
    )

    fig.update_xaxes(title_text="Time Step", row=1, col=1)
    fig.update_xaxes(title_text="Time Step", row=1, col=2)
    fig.update_xaxes(title_text="Time Step", row=2, col=1)
    fig.update_xaxes(title_text="X Position", row=2, col=2)

    fig.update_yaxes(title_text="Position", row=1, col=1)
    fig.update_yaxes(title_text="Velocity", row=1, col=2)
    fig.update_yaxes(title_text="Error", row=2, col=1)
    fig.update_yaxes(title_text="Y Position", row=2, col=2)

    return fig

def create_uncertainty_plot(measurements, filtered_positions, true_trajectory,
                           covariance_history, plot_interval=5):
    """
    Create uncertainty visualization with confidence ellipses
    """
    fig = go.Figure()

    # Plot GPS measurements
    fig.add_trace(
        go.Scatter(x=measurements[:, 0], y=measurements[:, 1],
                  mode='markers', name='GPS Measurements',
                  marker=dict(color='gray', size=8, opacity=0.3))
    )

    # Plot true trajectory
    fig.add_trace(
        go.Scatter(x=true_trajectory[:, 0], y=true_trajectory[:, 1],
                  mode='lines+markers', name='True Trajectory',
                  line=dict(color='green', width=3), marker=dict(size=10))
    )

    # Plot filtered positions with uncertainty ellipses
    n_plots = len(measurements)
    for i in range(0, n_plots, plot_interval):
        pos = filtered_positions[i]
        cov = covariance_history[i][:2, :2]

        # Calculate ellipse parameters
        eigenvalues, eigenvectors = np.linalg.eig(cov)
        angle = np.arctan2(eigenvectors[1, 0], eigenvectors[0, 0])

        # Confidence ellipse (95% confidence = 2.447 sigma for 2D)
        chi2_95 = 5.991  # 95% confidence for 2D
        a = np.sqrt(chi2_95 * eigenvalues[0])
        b = np.sqrt(chi2_95 * eigenvalues[1])

        # Generate ellipse points
        theta = np.linspace(0, 2*np.pi, 50)
        ellipse_x = a * np.cos(theta)
        ellipse_y = b * np.sin(theta)

        # Rotate ellipse
        R = np.array([[np.cos(angle), -np.sin(angle)],
                     [np.sin(angle), np.cos(angle)]])
        ellipse_points = np.array([ellipse_x, ellipse_y]).T
        rotated_points = ellipse_points @ R.T
        ellipse_x_rot = rotated_points[:, 0] + pos[0]
        ellipse_y_rot = rotated_points[:, 1] + pos[1]

        fig.add_trace(
            go.Scatter(x=ellipse_x_rot, y=ellipse_y_rot,
                      mode='lines', name='Confidence Region' if i == 0 else '',
                      line=dict(color='red', width=1, dash='dash'),
                      showlegend=(i == 0),
                      opacity=0.7)
        )

    # Add filtered trajectory
    fig.add_trace(
        go.Scatter(x=filtered_positions[:, 0], y=filtered_positions[:, 1],
                  mode='lines+markers', name='Filtered Trajectory',
                  line=dict(color='red', width=2), marker=dict(size=6))
    )

    fig.update_layout(
        title="Kalman Filter Uncertainty Visualization",
        xaxis_title="X Position",
        yaxis_title="Y Position",
        height=600,
        showlegend=True
    )

    return fig

def main():
    """
    Main function demonstrating the Kalman filter with synthetic data
    """
    # Parameters
    dt = 1.0
    num_steps = 100
    process_noise_std = 0.5
    measurement_noise_std = 10.0

    # Generate true trajectory (constant velocity)
    true_trajectory = []
    px, py = 0.0, 0.0
    vx, vy = 2.0, 1.5

    for k in range(num_steps):
        # Add slight random acceleration to make it interesting
        ax = np.random.normal(0, process_noise_std * 0.1)
        ay = np.random.normal(0, process_noise_std * 0.1)

        vx += ax * dt
        vy += ay * dt
        px += vx * dt
        py += vy * dt

        true_trajectory.append([px, py, vx, vy])

    true_trajectory = np.array(true_trajectory)

    # Generate GPS measurements with noise
    measurements = generate_gps_data(true_trajectory, measurement_noise_std, dt)

    # Run Kalman filter
    filtered_states, filtered_positions = run_kalman_filter(
        measurements, dt, process_noise_std, measurement_noise_std
    )

    # Calculate covariance history for uncertainty visualization
    kf = KalmanFilter2D(dt=dt, process_noise_std=process_noise_std,
                       measurement_noise_std=measurement_noise_std)
    kf.initialize(measurements[0], initial_velocity=[0, 0])

    covariance_history = [kf.get_covariance()]
    for measurement in measurements:
        kf.predict()
        kf.update(measurement)
        covariance_history.append(kf.get_covariance())

    # Create visualizations
    fig1 = create_comparison_plot(measurements, true_trajectory,
                                  filtered_positions, covariance_history)

    fig2 = create_uncertainty_plot(measurements, filtered_positions,
                                   true_trajectory, covariance_history,
                                   plot_interval=10)

    # Show the plots
    fig1.show()
    fig2.show()

    # Print performance metrics
    rmse_x = np.sqrt(np.mean((filtered_positions[:, 0] - true_trajectory[:, 0])**2))
    rmse_y = np.sqrt(np.mean((filtered_positions[:, 1] - true_trajectory[:, 1])**2))
    rmse_meas_x = np.sqrt(np.mean((measurements[:, 0] - true_trajectory[:, 0])**2))
    rmse_meas_y = np.sqrt(np.mean((measurements[:, 1] - true_trajectory[:, 1])**2))

    print("=" * 50)
    print("Kalman Filter Performance Summary")
    print("=" * 50)
    print(f"Measurement Noise Std: {measurement_noise_std:.2f}")
    print(f"Process Noise Std: {process_noise_std:.2f}")
    print(f"RMSE (Measurement - True) X: {rmse_meas_x:.2f}")
    print(f"RMSE (Measurement - True) Y: {rmse_meas_y:.2f}")
    print(f"RMSE (Filtered - True) X: {rmse_x:.2f}")
    print(f"RMSE (Filtered - True) Y: {rmse_y:.2f}")
    print(f"Improvement X: {((rmse_meas_x - rmse_x) / rmse_meas_x * 100):.1f}%")
    print(f"Improvement Y: {((rmse_meas_y - rmse_y) / rmse_meas_y * 100):.1f}%")
    print("=" * 50)

    return filtered_states, filtered_positions, measurements, true_trajectory

if __name__ == "__main__":
    filtered_states, filtered_positions, measurements, true_trajectory = main()
```
