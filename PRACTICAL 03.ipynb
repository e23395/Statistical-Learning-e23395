##**(b)**

**Analytical Solution**
$$\frac{dy}{dt} = 2\sqrt{y}$$

$$\frac{1}{\sqrt{y}} \, dy = 2 \, dt$$

$$\int y^{-1/2} \, dy = \int 2 \, dt$$

$$2\sqrt{y} = 2t + C$$

$$\sqrt{y} = t + \frac{C}{2}$$

Let $C_1 = \frac{C}{2}$. Using the initial condition $y(0) = y_0$:

$$\sqrt{y_0} = 0 + C_1 \implies C_1 = \sqrt{y_0}$$

$$\sqrt{y} = t + \sqrt{y_0}$$

$$y(t) = (t + \sqrt{y_0})^2$$


**Comparison with the Forward Euler Method**

For $y_0 > 0$: The exact solution grows quadratically as $y(t) = (t + \sqrt{y_0})^2$. The Forward Euler numerical solution will approximate this parabolic curve. Because the derivative $2\sqrt{y}$ is an increasing function (the solution is concave up), Forward Euler (which uses the slope at the beginning of each step) will consistently underestimate the true analytical solution.

**What happens when $y_0 = 0$?**

When $y_0 = 0$, the Forward Euler method gets trapped at $y = 0$ forever and completely fails to see the moving analytical solution $y(t) = t^2$. It purely tracks the trivial equilibrium solution $y(t) = 0$.

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def forward_euler(h, y0, t0=0, n_steps=20):
    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)
    t[0] = t0
    y[0] = y0

    for k in range(n_steps):
        t[k+1] = t[k] + h
        y[k+1] = y[k] + 2 * h * np.sqrt(y[k])

    return t, y

def exact_solution(t, y0):
    return (np.sqrt(y0) + t)**2

y0 = 0.2

# Part (c)(i-iii): Compute solutions for different step sizes
h1, n1 = 0.1, 20
h2, n2 = 0.05, 40
h3, n3 = 0.025, 80

t1, y1 = forward_euler(h1, y0, n_steps=n1)
t2, y2 = forward_euler(h2, y0, n_steps=n2)
t3, y3 = forward_euler(h3, y0, n_steps=n3)

# Exact solution
t_exact = np.linspace(0, 2.0, 200)
y_exact = exact_solution(t_exact, y0)

# Plot numerical solutions
fig1 = go.Figure()

fig1.add_trace(go.Scatter(
    x=t1, y=y1,
    mode='lines+markers',
    name=f'h = {h1}',
    line=dict(width=2)
))

fig1.add_trace(go.Scatter(
    x=t2, y=y2,
    mode='lines+markers',
    name=f'h = {h2}',
    line=dict(width=2)
))

fig1.add_trace(go.Scatter(
    x=t3, y=y3,
    mode='lines+markers',
    name=f'h = {h3}',
    line=dict(width=2)
))

fig1.add_trace(go.Scatter(
    x=t_exact, y=y_exact,
    mode='lines',
    name='Exact Solution',
    line=dict(color='red', width=2, dash='dash')
))

fig1.update_layout(
    title='Forward Euler Solutions for Different Step Sizes (y₀ = 0.2)',
    xaxis_title='t',
    yaxis_title='y',
    template='plotly_white',
    hovermode='x unified'
)

fig1.show()

# Compute r_k
def compute_r(y1, y2, y3):
    # y1: h, y2: h/2, y3: h/4
    # r_k = (y_k^(i) - y_{2k}^(ii)) / (y_{2k}^(ii) - y_{4k}^(iii))
    # where y_k^(i) corresponds to h, y_{2k}^(ii) to h/2, y_{4k}^(iii) to h/4

    # We need to compare at same t values
    # y1 at index k, y2 at index 2k, y3 at index 4k
    n = len(y1)  # number of points for h
    r = np.zeros(n)
    r[0] = 0  # r_0 = 0

    for k in range(1, n):
        denom = y2[2*k] - y3[4*k]
        if denom != 0:
            r[k] = (y1[k] - y2[2*k]) / denom
        else:
            r[k] = 0

    return r

r_values = compute_r(y1, y2, y3)

# Plot r_k
fig2 = go.Figure()

fig2.add_trace(go.Scatter(
    x=t1, y=r_values,
    mode='lines+markers',
    name='r_k',
    line=dict(color='blue', width=2),
    marker=dict(size=6)
))

fig2.add_hline(y=1, line_dash="dash", line_color="red",
               annotation_text="r = 1", annotation_position="top right")

fig2.update_layout(
    title='r_k vs t_k (h = 0.1, 0.05, 0.025)',
    xaxis_title='t',
    yaxis_title='r_k',
    template='plotly_white',
    hovermode='x unified'
)

fig2.show()

# Part (c): Repeat with halved step sizes
print("\nPart (c) - Halved Step Sizes:")
print("=" * 50)

h1_new, n1_new = 0.05, 40
h2_new, n2_new = 0.025, 80
h3_new, n3_new = 0.0125, 160

t1_new, y1_new = forward_euler(h1_new, y0, n_steps=n1_new)
t2_new, y2_new = forward_euler(h2_new, y0, n_steps=n2_new)
t3_new, y3_new = forward_euler(h3_new, y0, n_steps=n3_new)

# Compute r_k for new step sizes
r_new = compute_r(y1_new, y2_new, y3_new)

# Plot new r_k
fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=t1_new, y=r_new,
    mode='lines+markers',
    name='r_k (halved)',
    line=dict(color='green', width=2),
    marker=dict(size=6)
))

fig3.add_hline(y=1, line_dash="dash", line_color="red",
               annotation_text="r = 1", annotation_position="top right")

fig3.update_layout(
    title='r_k vs t_k (h = 0.05, 0.025, 0.0125)',
    xaxis_title='t',
    yaxis_title='r_k',
    template='plotly_white',
    hovermode='x unified'
)

fig3.show()
```

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def forward_euler_modified(h, y0, f, t0=0, n_steps=20):
    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)
    t[0] = t0
    y[0] = y0

    for k in range(n_steps):
        t[k+1] = t[k] + h
        y[k+1] = y[k] + h * f(y[k])

    return t, y

def exact_solution_linear(y0, t):
    return y0 * np.exp(2 * t)

def exact_solution_linear_plus(y0, t):
    return (1 + y0) * np.exp(2 * t) - 1

def compute_r(y1, y2, y3):
    n = len(y1)
    r = np.zeros(n)
    r[0] = 0

    for k in range(1, n):
        denom = y2[2*k] - y3[4*k]
        if denom != 0:
            r[k] = (y1[k] - y2[2*k]) / denom
        else:
            r[k] = 0

    return r

y0 = 0.2

# Choose one equation to test
# Option 1: dy/dt = 2y
def f1(y):
    return 2 * y

# Option 2: dy/dt = 2(1+y)
def f2(y):
    return 2 * (1 + y)

# Using dy/dt = 2y (you can change to f2 for the other equation)
f = f1
eq_name = "dy/dt = 2y"

print(f"Part (d): Modified equation {eq_name}")
print("=" * 50)

# Step sizes
h1, n1 = 0.1, 20
h2, n2 = 0.05, 40
h3, n3 = 0.025, 80

t1, y1 = forward_euler_modified(h1, y0, f, n_steps=n1)
t2, y2 = forward_euler_modified(h2, y0, f, n_steps=n2)
t3, y3 = forward_euler_modified(h3, y0, f, n_steps=n3)

# Exact solution
t_exact = np.linspace(0, 2.0, 200)
if f == f1:
    y_exact = exact_solution_linear(y0, t_exact)
else:
    y_exact = exact_solution_linear_plus(y0, t_exact)

# Plot numerical solutions
fig1 = go.Figure()

fig1.add_trace(go.Scatter(
    x=t1, y=y1,
    mode='lines+markers',
    name=f'h = {h1}',
    line=dict(width=2)
))

fig1.add_trace(go.Scatter(
    x=t2, y=y2,
    mode='lines+markers',
    name=f'h = {h2}',
    line=dict(width=2)
))

fig1.add_trace(go.Scatter(
    x=t3, y=y3,
    mode='lines+markers',
    name=f'h = {h3}',
    line=dict(width=2)
))

fig1.add_trace(go.Scatter(
    x=t_exact, y=y_exact,
    mode='lines',
    name='Exact Solution',
    line=dict(color='red', width=2, dash='dash')
))

fig1.update_layout(
    title=f'Forward Euler Solutions for {eq_name} (y₀ = {y0})',
    xaxis_title='t',
    yaxis_title='y',
    template='plotly_white',
    hovermode='x unified'
)

fig1.show()

# Compute r_k
r_values = compute_r(y1, y2, y3)

# Plot r_k
fig2 = go.Figure()

fig2.add_trace(go.Scatter(
    x=t1, y=r_values,
    mode='lines+markers',
    name='r_k',
    line=dict(color='blue', width=2),
    marker=dict(size=6)
))

fig2.add_hline(y=1, line_dash="dash", line_color="red",
               annotation_text="r = 1", annotation_position="top right")

fig2.update_layout(
    title=f'r_k vs t_k for {eq_name}',
    xaxis_title='t',
    yaxis_title='r_k',
    template='plotly_white',
    hovermode='x unified'
)

fig2.show()

# Repeat with halved step sizes
print("\nHalved Step Sizes:")
print("-" * 30)

h1_new, n1_new = 0.05, 40
h2_new, n2_new = 0.025, 80
h3_new, n3_new = 0.0125, 160

t1_new, y1_new = forward_euler_modified(h1_new, y0, f, n_steps=n1_new)
t2_new, y2_new = forward_euler_modified(h2_new, y0, f, n_steps=n2_new)
t3_new, y3_new = forward_euler_modified(h3_new, y0, f, n_steps=n3_new)

r_new = compute_r(y1_new, y2_new, y3_new)

fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=t1_new, y=r_new,
    mode='lines+markers',
    name='r_k (halved)',
    line=dict(color='green', width=2),
    marker=dict(size=6)
))

fig3.add_hline(y=1, line_dash="dash", line_color="red",
               annotation_text="r = 1", annotation_position="top right")

fig3.update_layout(
    title=f'r_k vs t_k for {eq_name} (halved step sizes)',
    xaxis_title='t',
    yaxis_title='r_k',
    template='plotly_white',
    hovermode='x unified'
)

fig3.show()

```

**Why is $r_k$ approximately constant?**
*   Method Dependent, Not ODE Dependent: The ratio $r_k$ depends on the order of the numerical method, not the specific equation $g(y)$.

*   Dominant Error Term: For any smooth function $g(y)$, the leading global error term is proportional to $h^1$. As $h \to 0$, higher-order errors ($h^2, h^3$) vanish, leaving a stable, constant error ratio across all time steps.

**Why does it take this particular value ($r_k \approx 2$)?**

*   First-Order Accuracy: Forward Euler is a first-order method ($p=1$).

*   Halving the Step Size: Since error scales linearly ($O(h)$), halving the step size halves the error.

*   $$\text{Numerator: } E(h) - E\left(\frac{h}{2}\right) \propto h - \frac{h}{2} = \frac{h}{2}$$$$\text{Denominator: } E\left(\frac{h}{2}\right) - E\left(\frac{h}{4}\right) \propto \frac{h}{2} - \frac{h}{4} = \frac{h}{4}$$$$r_k \approx \frac{h/2}{h/4} = 2$$

```python
import numpy as np
import plotly.graph_objects as go

def explicit_midpoint(h, y0, t0=0, t_end=20):
    """
    Explicit Midpoint method for dy/dt = sin^3(t)

    Parameters:
    h: step size
    y0: initial condition
    t0: initial time
    t_end: end time

    Returns:
    t: array of time values
    y: array of solution values
    """
    n_steps = int((t_end - t0) / h)

    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)

    t[0] = t0
    y[0] = y0

    def f(t):
        return np.sin(t)**3

    for k in range(n_steps):
        # First stage: half-step
        t_half = t[k] + h/2
        y_half = y[k] + (h/2) * f(t[k])

        # Second stage: full step using midpoint slope
        t[k+1] = t[k] + h
        y[k+1] = y[k] + h * f(t_half)

    return t, y

# Part (a): Solve with h = 0.1, y0 = -1
h = 0.1
y0 = -1
t_end = 20

t, y = explicit_midpoint(h, y0, t_end=t_end)

# Print first few values
print("Explicit Midpoint Method: dy/dt = sin^3(t), y(0) = -1")
print("h = 0.1, t ∈ [0, 20]")
print("-" * 50)
print("k\t t_k\t\t y_k")
print("-" * 50)
for k in range(min(10, len(t))):
    print(f"{k}\t {t[k]:.2f}\t\t {y[k]:.6f}")
print("...")
print(f"{len(t)-1}\t {t[-1]:.2f}\t\t {y[-1]:.6f}")

# Plot numerical solution
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=t,
    y=y,
    mode='lines',
    name=f'Midpoint Method (h={h})',
    line=dict(color='blue', width=2)
))

fig.add_trace(go.Scatter(
    x=t,
    y=y,
    mode='markers',
    name='Points',
    marker=dict(color='blue', size=3, opacity=0.5)
))

fig.update_layout(
    title=f'Explicit Midpoint Method: dy/dt = sin³(t), y(0) = {y0}',
    xaxis_title='t',
    yaxis_title='y',
    template='plotly_white',
    hovermode='x unified',
    showlegend=True
)

fig.show()

# Analytical solution for verification
# ∫ sin³(t) dt = -cos(t) + (1/3)cos³(t) + C
# y(t) = -cos(t) + (1/3)cos³(t) + C
# y(0) = -1 => C = -1 + 1 - 1/3 = -1/3
# y(t) = -cos(t) + (1/3)cos³(t) - 1/3

def exact_solution(t):
    return -np.cos(t) + (1/3) * np.cos(t)**3 - 1/3

# Plot comparison with exact solution
t_exact = np.linspace(0, 20, 1000)
y_exact = exact_solution(t_exact)

fig2 = go.Figure()

fig2.add_trace(go.Scatter(
    x=t,
    y=y,
    mode='lines',
    name=f'Midpoint Method (h={h})',
    line=dict(color='blue', width=2)
))

fig2.add_trace(go.Scatter(
    x=t_exact,
    y=y_exact,
    mode='lines',
    name='Exact Solution',
    line=dict(color='red', width=2, dash='dash')
))

fig2.update_layout(
    title='Comparison: Midpoint Method vs Exact Solution',
    xaxis_title='t',
    yaxis_title='y',
    template='plotly_white',
    hovermode='x unified'
)

fig2.show()

# Compute and plot error
error = y - exact_solution(t)

fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=t,
    y=error,
    mode='lines',
    name='Error',
    line=dict(color='red', width=2)
))

fig3.add_hline(y=0, line_dash="dash", line_color="black")

fig3.update_layout(
    title='Error: Numerical - Exact Solution',
    xaxis_title='t',
    yaxis_title='Error',
    template='plotly_white',
    hovermode='x unified'
)

fig3.show()

# Function documentation
print("\n" + "="*50)
print("Function Documentation:")
print("="*50)
print(explicit_midpoint.__doc__)
```

##**(b)**

**Analytical Solution**

The ODE is directly integrable since the right-hand side depends only on $t$:

$$\frac{dy}{dt} = \sin^3(t)$$Using the trigonometric identity $\sin^3(t) = \frac{3}{4}\sin(t) - \frac{1}{4}\sin(3t)$:

$$y(t) = \int \left( \frac{3}{4}\sin(t) - \frac{1}{4}\sin(3t) \right) dt = -\frac{3}{4}\cos(t) + \frac{1}{12}\cos(3t) + C$$

Apply the initial condition $y(0) = -1$:

$$-1 = -\frac{3}{4}(1) + \frac{1}{12}(1) + C \implies -1 = -\frac{8}{12} + C \implies C = -\frac{1}{3}$$

Thus, the exact analytical solution is:

$$y(t) = -\frac{3}{4}\cos(t) + \frac{1}{12}\cos(3t) - \frac{1}{3}$$

```python
import numpy as np
import plotly.graph_objects as go

def explicit_midpoint(h, y0, t0=0, t_end=20):
    n_steps = int((t_end - t0) / h)
    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)
    t[0] = t0
    y[0] = y0

    def f(t):
        return np.sin(t)**3

    for k in range(n_steps):
        t_half = t[k] + h/2
        y_half = y[k] + (h/2) * f(t[k])
        t[k+1] = t[k] + h
        y[k+1] = y[k] + h * f(t_half)

    return t, y

def compute_r(y1, y2, y3):
    n = len(y1)
    r = np.zeros(n)
    r[0] = 0

    for k in range(1, n):
        denom = y2[2*k] - y3[4*k]
        if denom != 0:
            r[k] = (y1[k] - y2[2*k]) / denom
        else:
            r[k] = 0

    return r

y0 = -1
t_end = 20

print("Part (b): Computing r_k for explicit midpoint method")
print("=" * 60)

# First set: h = 0.1, 0.05, 0.025
h1, n1 = 0.1, int(t_end/0.1)
h2, n2 = 0.05, int(t_end/0.05)
h3, n3 = 0.025, int(t_end/0.025)

t1, y1 = explicit_midpoint(h1, y0, t_end=t_end)
t2, y2 = explicit_midpoint(h2, y0, t_end=t_end)
t3, y3 = explicit_midpoint(h3, y0, t_end=t_end)

r_values = compute_r(y1, y2, y3)

fig1 = go.Figure()

fig1.add_trace(go.Scatter(
    x=t1,
    y=r_values,
    mode='lines+markers',
    name='r_k (h = 0.1, 0.05, 0.025)',
    line=dict(color='blue', width=2),
    marker=dict(size=4)
))

fig1.add_hline(y=4, line_dash="dash", line_color="red",
               annotation_text="r = 4 (2nd order)",
               annotation_position="top right")

fig1.update_layout(
    title='r_k vs t_k (h = 0.1, 0.05, 0.025)',
    xaxis_title='t',
    yaxis_title='r_k',
    template='plotly_white',
    hovermode='x unified',
    yaxis=dict(range=[0, 6])
)

fig1.show()

# Second set: h = 0.05, 0.025, 0.0125
h1_new, n1_new = 0.05, int(t_end/0.05)
h2_new, n2_new = 0.025, int(t_end/0.025)
h3_new, n3_new = 0.0125, int(t_end/0.0125)

t1_new, y1_new = explicit_midpoint(h1_new, y0, t_end=t_end)
t2_new, y2_new = explicit_midpoint(h2_new, y0, t_end=t_end)
t3_new, y3_new = explicit_midpoint(h3_new, y0, t_end=t_end)

r_new = compute_r(y1_new, y2_new, y3_new)

fig2 = go.Figure()

fig2.add_trace(go.Scatter(
    x=t1_new,
    y=r_new,
    mode='lines+markers',
    name='r_k (h = 0.05, 0.025, 0.0125)',
    line=dict(color='green', width=2),
    marker=dict(size=4)
))

fig2.add_hline(y=4, line_dash="dash", line_color="red",
               annotation_text="r = 4 (2nd order)",
               annotation_position="top right")

fig2.update_layout(
    title='r_k vs t_k (h = 0.05, 0.025, 0.0125)',
    xaxis_title='t',
    yaxis_title='r_k',
    template='plotly_white',
    hovermode='x unified',
    yaxis=dict(range=[0, 6])
)

fig2.show()

# Compute average r_k (excluding transient)
avg_r1 = np.mean(r_values[10:])
avg_r2 = np.mean(r_new[10:])
```

**What do you observe?**

*   The calculated ratio $r_k$ yields a value approximately equal to 4.0 across all time points.

*   Repeating the calculation with the halved set of step sizes yields the exact same behavior, with $r_k$ staying even tighter around 4.0.

**Why does it take this particular value?**

*   Second-Order Accuracy: The explicit midpoint method is a second-order ($p=2$) Runge-Kutta method, meaning its global accumulation error scales quadratically, $O(h^2)$.

*   Halving Effect: When the step size is cut in half ($h \to \frac{h}{2}$), the numerical error decreases by a factor of $2^2 = 4$.

*   $$\text{Numerator: } E(h) - E\left(\frac{h}{2}\right) \approx C \cdot h^2 - C \cdot \left(\frac{h}{2}\right)^2 = \frac{3}{4}Ch^2$$
$$\text{Denominator: } E\left(\frac{h}{2}\right) - E\left(\frac{h}{4}\right) \approx C \cdot \left(\frac{h}{2}\right)^2 - C \cdot \left(\frac{h}{4}\right)^2 = \frac{3}{16}Ch^2$$
$$r_k \approx \frac{\frac{3}{4}Ch^2}{\frac{3}{16}Ch^2} = 4$$

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def explicit_midpoint(h, y0, t0=0, t_end=20):
    n_steps = int((t_end - t0) / h)
    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)
    t[0] = t0
    y[0] = y0

    def f(t):
        return np.sin(t)**3

    for k in range(n_steps):
        t_half = t[k] + h/2
        y_half = y[k] + (h/2) * f(t[k])
        t[k+1] = t[k] + h
        y[k+1] = y[k] + h * f(t_half)

    return t, y

def exact_solution(t):
    return -np.cos(t) + (1/3) * np.cos(t)**3 - 1/3

y0 = -1
t_end = 100

print("Part (c): Exploring step sizes for t ∈ [0, 100]")
print("=" * 70)

# Test different step sizes
step_sizes = [1.0, 0.5, 0.2, 0.1, 0.05, 0.025]
colors = ['red', 'orange', 'yellow', 'green', 'blue', 'purple']

# Plot all solutions
fig1 = go.Figure()

for h, color in zip(step_sizes, colors):
    t, y = explicit_midpoint(h, y0, t_end=t_end)
    fig1.add_trace(go.Scatter(
        x=t,
        y=y,
        mode='lines',
        name=f'h = {h}',
        line=dict(color=color, width=2)
    ))

# Add exact solution (with fewer points for performance)
t_exact = np.linspace(0, t_end, 5000)
y_exact = exact_solution(t_exact)
fig1.add_trace(go.Scatter(
    x=t_exact,
    y=y_exact,
    mode='lines',
    name='Exact',
    line=dict(color='black', width=2, dash='dash')
))

fig1.update_layout(
    title='Midpoint Method: Different Step Sizes (t ∈ [0, 100])',
    xaxis_title='t',
    yaxis_title='y',
    template='plotly_white',
    hovermode='x unified',
    legend=dict(x=0.02, y=0.98)
)

fig1.show()

# Zoomed in plot for first 10 seconds
fig2 = go.Figure()

for h, color in zip(step_sizes, colors):
    t, y = explicit_midpoint(h, y0, t_end=10)
    fig2.add_trace(go.Scatter(
        x=t,
        y=y,
        mode='lines',
        name=f'h = {h}',
        line=dict(color=color, width=2)
    ))

t_exact_zoom = np.linspace(0, 10, 1000)
y_exact_zoom = exact_solution(t_exact_zoom)
fig2.add_trace(go.Scatter(
    x=t_exact_zoom,
    y=y_exact_zoom,
    mode='lines',
    name='Exact',
    line=dict(color='black', width=2, dash='dash')
))

fig2.update_layout(
    title='Zoom: First 10 Seconds',
    xaxis_title='t',
    yaxis_title='y',
    template='plotly_white',
    hovermode='x unified',
    legend=dict(x=0.02, y=0.98)
)

fig2.show()

# Compute errors for different step sizes
print("\nError Analysis:")
print("-" * 70)
print("h\t\t Max Error\t\t Mean Error")
print("-" * 70)

for h in step_sizes[:5]:
    t, y = explicit_midpoint(h, y0, t_end=t_end)
    y_exact_at_t = exact_solution(t)
    max_error = np.max(np.abs(y - y_exact_at_t))
    mean_error = np.mean(np.abs(y - y_exact_at_t))
    print(f"{h:.3f}\t\t {max_error:.6f}\t\t {mean_error:.6f}")

# Plot error vs step size
h_values = step_sizes[:5]
max_errors = []
mean_errors = []

for h in h_values:
    t, y = explicit_midpoint(h, y0, t_end=t_end)
    y_exact_at_t = exact_solution(t)
    max_errors.append(np.max(np.abs(y - y_exact_at_t)))
    mean_errors.append(np.mean(np.abs(y - y_exact_at_t)))

fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=h_values,
    y=max_errors,
    mode='lines+markers',
    name='Max Error',
    line=dict(color='red', width=2),
    marker=dict(size=10)
))

fig3.add_trace(go.Scatter(
    x=h_values,
    y=mean_errors,
    mode='lines+markers',
    name='Mean Error',
    line=dict(color='blue', width=2),
    marker=dict(size=10)
))

fig3.update_layout(
    title='Error vs Step Size (log scale)',
    xaxis_title='h',
    yaxis_title='Error',
    xaxis_type='log',
    yaxis_type='log',
    template='plotly_white'
)

fig3.show()

```

**Minimum step size needed:**

Upper bound: $h$ must be strictly less than 1.0 to visually see the oscillations, but it needs to be $h \le 0.1$ to accurately capture the true amplitude and features of the solution.

The forcing term $\sin^3(t)$ has a fundamental period of $2\pi \approx 6.28$ seconds. According to the Nyquist sampling criterion, you need at least 2 points per cycle ($h < 3.14$) just to detect an oscillation, but resolving the smooth, non-linear curvature of $\sin^3(t)$ requires significantly finer grid spacing.

**Oscillatory Forcing Term $\sin^3(t)$ vs. Step Size $h$**

*   The term $\sin^3(t)$ introduces rapid periodic oscillations with a wave period of $2\pi \approx 6.28$. If the step size $h$ is chosen too large (e.g., $h \ge 1.0$), the numerical solver samples the wave too sparsely, missing peaks and troughs entirely.

*   To properly resolve the shape of this continuous wave instead of a distorted, aliased approximation, you need multiple grid points per cycle, dictating a practical upper bound of $h \le 0.1$.

**Step Size $h$ vs. Numerical Accuracy**

*   The explicit midpoint method has a global truncation error of $O(h^2)$. Cutting the step size $h$ in half increases accuracy by a factor of 4.

*   When $h$ drops below the threshold required by the forcing frequency ($h \le 0.1$), the numerical solution transitions from a chaotic, poorly sampled approximation to a highly accurate representation matching the analytical path.

**Accuracy of the Numerical Solution over $t \in [0, 100]$**

*   Over long time horizons like $t = 100$, local truncation errors accumulate at each step, causing a global phase lag or phase drift where the numerical wave shifts out of sync with the exact solution.

*   A step size that works adequately for a short duration (like $t \in [0, 20]$) will degrade over $t \in [0, 100]$. A very fine step size ($h \le 0.1$ or smaller) is required to suppress this error accumulation and preserve accuracy through the entire interval.
