```python
from google.colab import files
import pandas as pd
import numpy as np
import io
import logging
from datetime import datetime
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import scipy.stats as stats
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
import warnings
warnings.filterwarnings('ignore')

# Advanced logging framework configuration
logging.basicConfig(level=logging.INFO, format='%(asctime)s - [%(levelname)s] - %(message)s')
logger = logging.getLogger("PremiumDataEngine")


class VisualStudio:
    """
    Enterprise-ready visualization suite delivering embedded analytics
    using responsive styling and customized modern color palettes.
    """

    @staticmethod
    def render_bar(dataframe, metric_x, metric_y=None, chart_title="Bar Graph", label_x=None, label_y=None):
        """Generates a standalone HTML wrapped bar visualization with a refined Indigo/Teal theme."""
        try:
            if dataframe is None or dataframe.empty:
                raise ValueError("Source data structure cannot be empty.")

            fig = px.bar(
                dataframe, x=metric_x, y=metric_y, title=chart_title,
                labels={metric_x: label_x or metric_x, metric_y: label_y or metric_y},
                color_discrete_sequence=px.colors.sequential.Cividis,
                template="plotly_white"
            )
            fig.update_layout(
                title_font_size=16,
                margin=dict(l=40, r=40, t=50, b=40),
                paper_bgcolor='#F8F9FA',
                plot_bgcolor='#FFFFFF'
            )
            return fig.to_html(full_html=False)
        except Exception as error:
            logger.error(f"Failed to generate bar chart trace: {error}")
            return f"<div class='error'>Visualization Render Error: {error}</div>"

    @staticmethod
    def render_pie(dataframe, slice_labels, slice_values, chart_title="Proportional Partition Display"):
        """Compiles a responsive proportional partition display with a muted slate/blue palette."""
        try:
            if dataframe is None or dataframe.empty:
                raise ValueError("Source data structure cannot be empty.")

            fig = px.pie(
                dataframe, names=slice_labels, values=slice_values, title=chart_title,
                color_discrete_sequence=px.colors.qualitative.Prism,
                template="plotly_white"
            )
            fig.update_traces(textposition='inside', textinfo='percent+label')
            fig.update_layout(paper_bgcolor='#F8F9FA')
            return fig.to_html(full_html=False)
        except Exception as error:
            logger.error(f"Failed to generate pie chart trace: {error}")
            return f"<div class='error'>Visualization Render Error: {error}</div>"

    @staticmethod
    def render_histogram(dataframe, numerical_target, bucket_size=25, chart_title="Frequency Distribution", label_x=None):
        """Constructs an independent frequency histogram mapping structured densities using slate teal."""
        try:
            if dataframe is None or dataframe.empty:
                raise ValueError("Source data structure cannot be empty.")

            fig = px.histogram(
                dataframe, x=numerical_target, nbins=bucket_size, title=chart_title,
                labels={numerical_target: label_x or numerical_target},
                color_discrete_sequence=['#3A6B73'],
                template="plotly_white"
            )
            fig.update_layout(
                yaxis_title="Absolute Count",
                bargap=0.05,
                paper_bgcolor='#F8F9FA',
                plot_bgcolor='#FFFFFF'
            )
            return fig.to_html(full_html=False)
        except Exception as error:
            logger.error(f"Failed to generate histogram trace: {error}")
            return f"<div class='error'>Visualization Render Error: {error}</div>"


class DataSanitizer:
    """
    Robust analytical backend handling clean file ingestions, programmatic mapping,
    advanced validation structures, and dataset transformation engines.
    """

    def __init__(self):
        """Instantiates internal variables for clean isolated data states."""
        self.working_df = None
        self.continuous_features = []
        self.nominal_features = []
        self.immutable_backup = None

    def execute_local_upload(self):
        """Prompts immediate local filesystem imports securely via stream buffers."""
        try:
            print("Please select your local CSV dataset for ingestion:")
            imported_files = files.upload()

            if not imported_files:
                logger.warning("File upload context was bypassed by user.")
                return None

            for unique_id in imported_files.keys():
                self.working_df = pd.read_csv(io.BytesIO(imported_files[unique_id]))
                self.immutable_backup = self.working_df.copy()
                logger.info(f"Successfully structuralized dataset: {unique_id}")
                logger.info(f"Base Matrix Footprint: {self.working_df.shape[0]} rows x {self.working_df.shape[1]} features")
                self._map_feature_spaces()
                return self.working_df
        except Exception as error:
            logger.error(f"Critical execution error during file streaming ingestion: {error}")
            return None

    def _map_feature_spaces(self):
        """Dynamically partitions processing columns into concrete structural spaces."""
        if self.working_df is not None:
            self.continuous_features = self.working_df.select_dtypes(include=[np.number]).columns.tolist()
            self.nominal_features = self.working_df.select_dtypes(include=['object', 'category']).columns.tolist()
        else:
            logger.warning("Feature space partitioning invoked on an empty dataframe reference.")

    def _clean_malformed_placeholders(self):
        """Transforms structural text typos into unified native numpy NaN architectures."""
        noise_strings = ['?', 'n/a', 'NULL', '', 'None', 'null', 'NaN', 'nan', 'unknown', 'UNKNOWN', ' ']
        if self.working_df is not None:
            self.working_df = self.working_df.replace(noise_strings, np.nan)
            for feature in self.nominal_features:
                if self.working_df[feature].dtype == 'object':
                    self.working_df[feature] = self.working_df[feature].astype(str).str.strip()
            self._map_feature_spaces()

    def _execute_type_casting(self):
        """Evaluates textual types to extract implicit continuous elements automatically."""
        if self.working_df is None:
            return

        for active_col in self.working_df.columns:
            if self.working_df[active_col].dtype == 'object':
                try:
                    parsed_series = pd.to_numeric(self.working_df[active_col], errors='coerce')
                    if not parsed_series.isna().all() and parsed_series.nunique() > 1:
                        self.working_df[active_col] = parsed_series
                        logger.info(f"Automatic casting optimizer remapped '{active_col}' column to Float/Int space.")
                except Exception as ex:
                    logger.debug(f"Type conversion skipped for attribute {active_col}: {ex}")
        self._map_feature_spaces()

    def display_metrics_dashboard(self):
        """Prints a completely distinct, structural overview summary detailing feature matrices."""
        if self.working_df is None:
            logger.error("Global dataframe object has not been loaded yet.")
            return

        print("\n" + "═"*70)
        print(f" PRODUCTION METRICS RUNTIME AUDIT REPORT - {datetime.now().strftime('%Y-%m-%d %H:%M')}")
        print("═"*70)
        print(f" Matrix Rows      : {self.working_df.shape[0]:,}")
        print(f" Matrix Features  : {self.working_df.shape[1]:,}")
        print(f" Continuous Count : {len(self.continuous_features)} attributes mapped")
        print(f" Nominal Count    : {len(self.nominal_features)} attributes mapped")
        print("─"*70)

        print("\n📋 PREVIEW OF FIRST 20 RECORDS:")
        display(self.working_df.head(20))

        if self.continuous_features:
            print("\n📈 CONTINUOUS ATTRIBUTE DESCRIPTIVE STATS:")
            display(self.working_df[self.continuous_features].describe().T)

        print("\n🗂️ CATEGORICAL SPACE STRUCTURAL DENSITY:")
        for target_key in self.nominal_features:
            null_count = self.working_df[target_key].isna().sum()
            print(f"  ▪️ Field [{target_key:15}] -> Unique Keys: {self.working_df[target_key].nunique():4} | Nulls Present: {null_count}")
        print("═"*70 + "\n")

    def resolve_empty_records(self, configuration='mean', set_constant=None):
        """Implements high-fidelity continuous mapping and localized modal patches across array indices."""
        if self.working_df is None:
            logger.error("Aborted: Dataframe object reference is unassigned.")
            return None

        logger.info(f"Initiating imputation sequence using configuration style: [{configuration}]")
        self._clean_malformed_placeholders()
        self._execute_type_casting()

        for item in self.working_df.columns:
            null_sum = self.working_df[item].isna().sum()
            if null_sum > 0:
                try:
                    if item in self.continuous_features:
                        if configuration == 'mean':
                            self.working_df[item].fillna(self.working_df[item].mean(), inplace=True)
                        elif configuration == 'median':
                            self.working_df[item].fillna(self.working_df[item].median(), inplace=True)
                        elif configuration == 'mode':
                            fallback = self.working_df[item].mode()[0] if not self.working_df[item].mode().empty else 0
                            self.working_df[item].fillna(fallback, inplace=True)
                        elif configuration == 'constant' and set_constant is not None:
                            self.working_df[item].fillna(set_constant, inplace=True)
                    else:
                        if configuration in ['mean', 'median', 'mode']:
                            fallback_text = self.working_df[item].mode()[0] if not self.working_df[item].mode().empty else 'Unknown'
                            self.working_df[item].fillna(fallback_text, inplace=True)
                        elif configuration == 'constant' and set_constant is not None:
                            self.working_df[item].fillna(set_constant, inplace=True)
                    logger.info(f"  Fixed missing contexts inside [{item}]: Resolved {null_sum} elements.")
                except Exception as error:
                    logger.error(f"  Failed imputation patch on field {item}: {error}")

        logger.info(f"Pipeline Check: Residual matrix null value fields: {self.working_df.isna().sum().sum()}")
        return self.working_df

    def prune_redundant_records(self):
        """Identifies and purges duplicate rows within the active dataframe."""
        if self.working_df is None:
            return None
        pre_count = len(self.working_df)
        self.working_df.drop_duplicates(inplace=True)
        post_count = pre_count - len(self.working_df)
        logger.info(f"Deduplication Engine: Removed {post_count} identical records from active matrix arrays.")
        return self.working_df

    def process_statistical_anomalies(self, feature_scope=None, operation='flag', bound_coefficient=1.5):
        """Executes Interquartile-based variance tracing to flag anomalies or drop extreme records."""
        if self.working_df is None:
            return None
        if feature_scope is None:
            feature_scope = self.continuous_features

        composite_outlier_vector = pd.Series([False] * len(self.working_df), name='anomaly_detected', index=self.working_df.index)

        for field in feature_scope:
            if field in self.continuous_features:
                try:
                    first_pct = self.working_df[field].quantile(0.25)
                    third_pct = self.working_df[field].quantile(0.75)
                    inter_spread = third_pct - first_pct
                    lower_threshold = first_pct - bound_coefficient * inter_spread
                    upper_threshold = third_pct + bound_coefficient * inter_spread

                    tracked_anomalies = (self.working_df[field] < lower_threshold) | (self.working_df[field] > upper_threshold)
                    composite_outlier_vector |= tracked_anomalies
                except Exception as error:
                    logger.error(f"Error executing statistical anomaly tracing for {field}: {error}")

        if operation == 'flag':
            self.working_df['anomaly_detected'] = composite_outlier_vector
            logger.info(f"Anomaly Engine: Isolated {composite_outlier_vector.sum()} instances matching variance limits.")
        elif operation == 'delete':
            initial_len = len(self.working_df)
            self.working_df = self.working_df[~composite_outlier_vector]
            logger.info(f"Anomaly Engine: Cleaned {initial_len - len(self.working_df)} anomalous records from processing scope.")

        return self.working_df

    def drop_explicit_rows(self, raw_comma_indices):
        """Drops explicit dataset records matching user provided numerical indices."""
        if self.working_df is None: return None
        try:
            parsed_targets = [int(idx.strip()) for idx in str(raw_comma_indices).split(',') if idx.strip().isdigit()]
            initial_volume = len(self.working_df)
            self.working_df = self.working_df.drop(parsed_targets, errors='ignore')
            logger.info(f"Pruned row indices: {parsed_targets}. Removed {initial_volume - len(self.working_df)} records.")
            return self.working_df
        except Exception as error:
            logger.error(f"Failed manual row exclusion: {error}")
            return self.working_df

    def drop_explicit_columns(self, raw_comma_fields):
        """Drops explicit columns matching user provided labels from the processing pipeline."""
        if self.working_df is None: return None
        try:
            extracted_keys = [label.strip() for label in str(raw_comma_fields).split(',')]
            valid_drops = [label for label in extracted_keys if label in self.working_df.columns]
            self.working_df = self.working_df.drop(columns=valid_drops)
            logger.info(f"Purged data columns from environment workspace: {valid_drops}")
            self._map_feature_spaces()
            return self.working_df
        except Exception as error:
            logger.error(f"Failed manual attribute clearance: {error}")
            return self.working_df

    def scale_numerical_spaces(self, methodology='minmax', targeted_features=None):
        """Applies mathematical transformations to numerical ranges using distinct scikit architectures."""
        if self.working_df is None: return pd.DataFrame()
        if targeted_features is None: targeted_features = self.continuous_features

        targeted_features = [feat for feat in targeted_features if feat in self.continuous_features]
        if not targeted_features:
            logger.warning("No continuous numeric fields available for feature scaling pipelines.")
            return pd.DataFrame()

        engine_variants = {
            'minmax': MinMaxScaler(),
            'standard': StandardScaler(),
            'robust': RobustScaler()
        }

        selected_transformer = engine_variants.get(methodology.lower())
        if not selected_transformer:
            raise ValueError(f"Methodology variant invalid. Choose from: {list(engine_variants.keys())}")

        try:
            transformed_matrix = selected_transformer.fit_transform(self.working_df[targeted_features])
            scaled_dataframe = pd.DataFrame(
                transformed_matrix,
                columns=[f"scaled_{feat}" for feat in targeted_features],
                index=self.working_df.index
            )
            logger.info(f"Transformation Pipeline: Scaled [{len(targeted_features)}] continuous parameters using {methodology.upper()}.")
            return scaled_dataframe
        except Exception as error:
            logger.error(f"Failed continuous domain normalisation: {error}")
            return pd.DataFrame()

    def encode_categorical_spaces(self, mapping_style='onehot', targeted_features=None):
        """Converts string classifications into explicit numeric arrays using independent configurations."""
        if self.working_df is None: return pd.DataFrame()
        if targeted_features is None: targeted_features = self.nominal_features

        targeted_features = [feat for feat in targeted_features if feat in self.nominal_features]
        if not targeted_features:
            logger.warning("No categorical object categories available for structural mapping pipelines.")
            return pd.DataFrame()

        try:
            if mapping_style.lower() == 'onehot':
                processed_matrix = pd.get_dummies(self.working_df[targeted_features], prefix=targeted_features, dtype=float)
                logger.info(f"Transformation Pipeline: One-hot encoded {len(targeted_features)} categorical factors.")

            elif mapping_style.lower() == 'ordinal':
                matrix_encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
                transformed_array = matrix_encoder.fit_transform(self.working_df[targeted_features].astype(str))
                processed_matrix = pd.DataFrame(transformed_array, columns=targeted_features, index=self.working_df.index)
                logger.info(f"Transformation Pipeline: Ordinal mapped {len(targeted_features)} categorical spaces.")

            elif mapping_style.lower() == 'uniform':
                processed_matrix = pd.DataFrame(index=self.working_df.index)
                for column_key in targeted_features:
                    unique_cardinality = self.working_df[column_key].nunique()
                    if unique_cardinality > 1:
                        local_encoder = OrdinalEncoder()
                        raw_codes = local_encoder.fit_transform(self.working_df[column_key].fillna('Missing').astype(str).values.reshape(-1, 1))
                        uniform_distribution = (raw_codes - raw_codes.min()) / (raw_codes.max() - raw_codes.min())
                        processed_matrix[f"uniform_{column_key}"] = uniform_distribution.flatten()
                logger.info(f"Transformation Pipeline: Applied uniform min-max variance distribution to encoded indices.")
            else:
                raise ValueError("Encoding choice invalid. Choose from 'onehot', 'ordinal', 'uniform'")

            return processed_matrix
        except Exception as error:
            logger.error(f"Failed descriptive domain matrix encoding: {error}")
            return pd.DataFrame()

    def stitch_transformed_matrices(self, frame_numerical=None, frame_categorical=None):
        """Stitches isolated engineered arrays side-by-side with the master workspace."""
        if self.working_df is None: return None
        compiled_output = self.working_df.copy()

        if frame_numerical is not None and not frame_numerical.empty:
            compiled_output = pd.concat([compiled_output, frame_numerical], axis=1)
        if frame_categorical is not None and not frame_categorical.empty:
            compiled_output = pd.concat([compiled_output, frame_categorical], axis=1)

        logger.info(f"Structural Stitch Complete: Integrated feature payload dimension metrics are now: {compiled_output.shape}")
        return compiled_output

    def generate_univariate_grids(self, tracking_fields=None):
        """Constructs a customized interactive univariate visualization using sophisticated soft tones."""
        if self.working_df is None or self.working_df.empty: return
        if tracking_fields is None: tracking_fields = self.continuous_features[:3]

        for item in tracking_fields:
            if item not in self.working_df.columns: continue
            try:
                canvas = make_subplots(
                    rows=1, cols=3,
                    subplot_titles=[f"Violin Plot Map", f"Observation Spread Grid", f"Relative Bin Target Density"]
                )

                canvas.add_trace(go.Violin(y=self.working_df[item], name=item, box_visible=True, line_color='#4A7BB0', meanline_visible=True), row=1, col=1)
                canvas.add_trace(go.Scatter(x=self.working_df.index, y=self.working_df[item], mode='markers', marker=dict(size=4, opacity=0.5, color='#D98880')), row=1, col=2)
                canvas.add_trace(go.Histogram(x=self.working_df[item], nbinsx=25, marker_color='#52BE80'), row=1, col=3)

                canvas.update_layout(
                    height=400,
                    title_text=f"Multi-Panel Statistical Analysis: Attribute Space [{item}]",
                    showlegend=False,
                    template="plotly_white",
                    paper_bgcolor='#F8F9FA',
                    plot_bgcolor='#FFFFFF'
                )
                canvas.show()
            except Exception as error:
                logger.error(f"Failed plot compilation steps for attribute target {item}: {error}")

    def identify_bivariate_relationships(self, feature_alpha, feature_beta):
        """Evaluates interactions dynamically between feature configurations and plots customized charts."""
        if self.working_df is None: return
        if feature_alpha not in self.working_df.columns or feature_beta not in self.working_df.columns:
            logger.error(f"Key mapping error: Targets '{feature_alpha}' or '{feature_beta}' do not exist inside data context.")
            return

        space_type_a = 'numeric' if feature_alpha in self.continuous_features else 'categorical'
        space_type_b = 'numeric' if feature_beta in self.continuous_features else 'categorical'

        try:
            if space_type_a == 'numeric' and space_type_b == 'numeric':
                current_r = self.working_df[feature_alpha].corr(self.working_df[feature_beta])
                canvas = px.scatter(
                    self.working_df, x=feature_alpha, y=feature_beta, trendline='ols',
                    title=f"Continuous System Mapping: {feature_alpha} vs {feature_beta} (Pearson r: {current_r:.4f})",
                    color_discrete_sequence=['#E67E22'], template="plotly_white"
                )
                canvas.update_layout(paper_bgcolor='#F8F9FA', plot_bgcolor='#FFFFFF')
                canvas.show()

            elif (space_type_a == 'numeric' and space_type_b == 'categorical') or (space_type_a == 'categorical' and space_type_b == 'numeric'):
                continuous_target = feature_alpha if space_type_a == 'numeric' else feature_beta
                categorical_target = feature_beta if space_type_a == 'numeric' else feature_alpha
                canvas = px.box(
                    self.working_df, x=categorical_target, y=continuous_target, points='outliers',
                    title=f"Conditional Distribution Profile: {continuous_target} structured across [{categorical_target}]",
                    color_discrete_sequence=['#16A085'], template="plotly_white"
                )
                canvas.update_layout(paper_bgcolor='#F8F9FA', plot_bgcolor='#FFFFFF')
                canvas.show()

            else:
                cross_distribution = pd.crosstab(self.working_df[feature_alpha], self.working_df[feature_beta])
                canvas = px.bar(
                    cross_distribution, barmode='group',
                    title=f"Categorical Frequency Distribution: Grouping [{feature_alpha}] by [{feature_beta}]",
                    color_discrete_sequence=px.colors.qualitative.Pastel, template="plotly_white"
                )
                canvas.update_layout(paper_bgcolor='#F8F9FA', plot_bgcolor='#FFFFFF')
                canvas.show()
        except Exception as error:
            logger.error(f"Failed runtime execution processing bivariate trend elements: {error}")

    def display_density_distribution(self, label_field, visual_depth=10):
        """Builds an isolated frequency distribution plot displaying clean annotations on light themes."""
        if self.working_df is None or label_field not in self.working_df.columns: return

        try:
            frequency_counts = self.working_df[label_field].value_counts().head(visual_depth)
            proportional_shares = (frequency_counts / len(self.working_df) * 100).round(2)

            canvas = go.Figure(data=[
                go.Bar(
                    x=frequency_counts.index.astype(str), y=frequency_counts.values,
                    text=[f"Obs: {v}<br>Share: {p}%" for v, p in zip(frequency_counts.values, proportional_shares)],
                    textposition='outside',
                    marker_color='#7FB3D5'
                )
            ])
            canvas.update_layout(
                title=f"Volume & Share Frequency Discovery Log: {label_field}",
                xaxis_title=f"Categorical Elements ({label_field})",
                yaxis_title="Total Rows Detected",
                template="plotly_white",
                paper_bgcolor='#F8F9FA',
                plot_bgcolor='#FFFFFF'
            )
            canvas.show()
        except Exception as error:
            logger.error(f"Failed structural trace profiling visualization: {error}")

    def render_comprehensive_heatmap(self):
        """Computes statistical cross-associations and renders a well-calibrated light heatmap."""
        if self.working_df is None: return

        composite_keys = self.continuous_features + self.nominal_features
        dimension_size = len(composite_keys)

        if dimension_size == 0:
            logger.warning("Empty vector metrics profile space configuration.")
            return

        correlation_matrix = np.zeros((dimension_size, dimension_size))

        for row_idx, key_x in enumerate(composite_keys):
            for col_idx, key_y in enumerate(composite_keys):
                if row_idx == col_idx:
                    correlation_matrix[row_idx, col_idx] = 1.0
                elif row_idx > col_idx:
                    correlation_matrix[row_idx, col_idx] = correlation_matrix[col_idx, row_idx]
                else:
                    try:
                        x_is_numeric = key_x in self.continuous_features
                        y_is_numeric = key_y in self.continuous_features

                        if x_is_numeric and y_is_numeric:
                            val = self.working_df[key_x].corr(self.working_df[key_y])
                            correlation_matrix[row_idx, col_idx] = val if not np.isnan(val) else 0.0
                        elif not x_is_numeric and not y_is_numeric:
                            cross_tab = pd.crosstab(self.working_df[key_x], self.working_df[key_y])
                            if cross_tab.size > 0 and cross_tab.sum().sum() > 0:
                                chi2_val = stats.chi2_contingency(cross_tab)[0]
                                grand_sum = cross_tab.sum().sum()
                                degrees_freedom = min(cross_tab.shape) - 1
                                correlation_matrix[row_idx, col_idx] = np.sqrt(chi2_val / (grand_sum * degrees_freedom)) if degrees_freedom > 0 else 0.0
                            else:
                                correlation_matrix[row_idx, col_idx] = 0.0
                        else:
                            numeric_field = key_x if x_is_numeric else key_y
                            categorical_field = key_y if x_is_numeric else key_x

                            grouped = self.working_df.groupby(categorical_field)[numeric_field].mean()
                            if len(grouped) > 1 and self.working_df[numeric_field].std() > 0:
                                dummy_frame = pd.get_dummies(self.working_df[categorical_field], drop_first=True)
                                calculated_coefficient = self.working_df[numeric_field].corr(dummy_frame.iloc[:, 0]) if dummy_frame.shape[1] > 0 else 0.0
                                correlation_matrix[row_idx, col_idx] = abs(calculated_coefficient) if not np.isnan(calculated_coefficient) else 0.0
                            else:
                                correlation_matrix[row_idx, col_idx] = 0.0
                    except Exception:
                        correlation_matrix[row_idx, col_idx] = 0.0

        canvas = px.imshow(
            correlation_matrix, x=composite_keys, y=composite_keys,
            color_continuous_scale='YlGnBu', text_auto='.2f',
            title="Unified Multi-Type Cross Association Grid Matrix",
            zmin=-1.0, zmax=1.0
        )
        canvas.update_layout(
            height=max(500, 35 * dimension_size),
            width=max(500, 35 * dimension_size),
            paper_bgcolor='#F8F9FA'
        )
        canvas.show()

    def plot_missingness_null_matrix(self):
        """[ADVANCED] Next-Level Feature: Plots an upgraded clean light null dependency map."""
        if self.working_df is None: return

        null_mask = self.working_df.isna().astype(float)
        columns_with_nulls = null_mask.columns[null_mask.sum() > 0].tolist()

        if not columns_with_nulls:
            logger.info("Advanced Visualizer: Clean dataset discovered; skipping null-missingness map.")
            return

        null_corr = null_mask[columns_with_nulls].corr().fillna(0.0)

        fig = px.imshow(
            null_corr, x=columns_with_nulls, y=columns_with_nulls,
            color_continuous_scale='Purples', text_auto='.2f',
            title="Advanced System Diagnostic: Missingness Null Correlation Profile Map"
        )
        fig.update_layout(template="plotly_white", paper_bgcolor='#F8F9FA')
        fig.show()

    def plot_skewness_normality_analytics(self):
        """[ADVANCED] Next-Level Feature: Returns a dynamic normality panel with high-visibility corporate tones."""
        if self.working_df is None or not self.continuous_features: return

        metrics_payload = []
        for feat in self.continuous_features:
            clean_vector = self.working_df[feat].dropna()
            if len(clean_vector) > 3:
                skew_coeff = stats.skew(clean_vector)
                kurtosis_coeff = stats.kurtosis(clean_vector)
                stat_sig = stats.shapiro(clean_vector.sample(min(len(clean_vector), 500)))[1]
                normality_verdict = "Normal" if stat_sig > 0.05 else "Skewed Non-Normal"

                metrics_payload.append({
                    "Feature Column": feat, "Skewness": skew_coeff,
                    "Kurtosis": kurtosis_coeff, "Normality Verdict": normality_verdict
                })

        metrics_df = pd.DataFrame(metrics_payload)

        fig = px.bar(
            metrics_df, x="Feature Column", y="Skewness", color="Normality Verdict",
            text_auto='.2f', title="Advanced Feature Diagnostic: Continuous Skewness Profile Panel",
            color_discrete_map={"Normal": "#58D68D", "Skewed Non-Normal": "#EC7063"}
        )
        fig.update_layout(template="plotly_white", yaxis_title="Calculated Skewness Coefficient", paper_bgcolor='#F8F9FA', plot_bgcolor='#FFFFFF')
        fig.show()

    def plot_high_dimensional_bubble_space(self, axis_x, axis_y, axis_size, axis_color):
        """[ADVANCED] Next-Level Feature: Maps multi-feature dynamics using a modern, soft-toned corporate style."""
        if self.working_df is None: return
        required_keys = [axis_x, axis_y, axis_size, axis_color]
        if not all(k in self.working_df.columns for k in required_keys):
            logger.error("Advanced Visualizer: Skipping multidimensional bubble map due to key mismatches.")
            return

        try:
            clean_scope = self.working_df[required_keys].dropna()

            fig = px.scatter(
                clean_scope, x=axis_x, y=axis_y, size=axis_size, color=axis_color,
                title=f"Advanced Multidimensional Space Mapping: {axis_x} vs {axis_y}",
                color_continuous_scale=px.colors.sequential.Mint,
                size_max=25, template="plotly_white"
            )
            fig.update_layout(paper_bgcolor='#F8F9FA', plot_bgcolor='#FFFFFF')
            fig.show()
        except Exception as err:
            logger.error(f"Failed high-dimensional canvas assembly: {err}")


if __name__ == "__main__":
    logger.info("Initializing Enterprise Engine Pipeline Verification...")

    # Instantiating production configuration class
    sanitizer = DataSanitizer()

    logger.info("Pulling standard dataset file stream into ingestion zone...")
    titanic_source_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

    try:
        # Ingest and cache baseline records
        sanitizer.working_df = pd.read_csv(titanic_source_url)
        sanitizer.immutable_backup = sanitizer.working_df.copy()
        sanitizer._map_feature_spaces()
        logger.info("Dataset safely cached into processing environment.")

        # 1. Run Standard Baseline Structural Visualizations
        sanitizer.display_metrics_dashboard()

        # 2. Run Advanced Missingness Diagnostic Graph
        sanitizer.plot_missingness_null_matrix()

        # Continue standard requirements flow
        sanitizer.resolve_empty_records(configuration='median')
        sanitizer.prune_redundant_records()
        sanitizer.process_statistical_anomalies(feature_scope=['Age', 'Fare'], operation='flag')

        # 3. Run Advanced Feature Distribution Assessment Graph
        sanitizer.plot_skewness_normality_analytics()

        # Feature Transformation Processing Steps
        extracted_nums = sanitizer.scale_numerical_spaces(methodology='minmax', targeted_features=['Age', 'Fare'])
        extracted_cats = sanitizer.encode_categorical_spaces(mapping_style='onehot', targeted_features=['Sex', 'Embarked'])
        merged_output = sanitizer.stitch_transformed_matrices(extracted_nums, extracted_cats)

        # 4. Run Original Required Analysis Visuals (Completely untouched)
        sanitizer.generate_univariate_grids(['Age', 'Fare'])
        sanitizer.identify_bivariate_relationships('Age', 'Fare')
        sanitizer.identify_bivariate_relationships('Sex', 'Survived')
        sanitizer.display_density_distribution('Pclass')
        sanitizer.render_comprehensive_heatmap()

        # 5. Run Multidimensional Multi-Factor Bubble Canvas
        sanitizer.plot_high_dimensional_bubble_space(axis_x='Age', axis_y='Fare', axis_size='Pclass', axis_color='Survived')

        studio_module = VisualStudio()
        logger.info("Automated testing workflow running securely. Engine stable and visually updated.")

    except Exception as general_error:
        logger.critical(f"Pipeline crashed during initialization check sequence: {general_error}")
```
